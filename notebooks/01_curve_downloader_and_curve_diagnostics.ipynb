```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from src.data.market_loader import MarketLoader

from src.curves.curve_snapshot import CurveSnapshot
from src.curves.base_curve import BaseCurve
```

```python
# downloading market curves
loader = MarketLoader()

market_curves = loader.market_loader_pipeline()
market_curves
```

```python
# summary statistics
market_curves.describe().T
```

```python
# curve snapshot of a given date
treasury_df = market_curves['treasury']

latest_date = treasury_df.index[-1]
latest_curve = treasury_df.iloc[-1]

snapshot = CurveSnapshot.snap_from_df_row(
    curve_name = 'treasury',
    as_of_date = latest_date,
    curve_row = latest_curve
)

display(snapshot.summary())

# linear interpolated base curve object
base_curve = BaseCurve(
    curve_snapshot = snapshot,
    interpolation_method = 'linear'
)

base_curve
```

```python
# downloading swap curves
swap_curves = loader.swap_loader_pipeline()
swap_curves
```

```python
# curve snapshot of a given date
usd_ois = swap_curves['usd_ois']
latest_date = usd_ois.index[-1]

latest_curve = usd_ois.iloc[-1]

swap_snapshot = CurveSnapshot.snap_from_df_row(
    curve_name="usd_ois",
    as_of_date=latest_date,
    curve_row=latest_curve
)

display(swap_snapshot.summary())

# linear interpolated base curve object
ois_curve = BaseCurve(
    curve_snapshot=swap_snapshot,
    interpolation_method="linear"
)

ois_curve
```
